# Análise Visual das Séries — Despesas Liquidadas

Exibe os gráficos de cada série gerados pelo experimento.

**Pré-requisito**: execute o notebook `01_experimento_geral.ipynb` com `gerar_graficos=True`.

```python
import sys
from pathlib import Path

ROOT = Path().resolve().parent
if str(ROOT) not in sys.path:
    sys.path.insert(0, str(ROOT))

from config import PLOTS_DIR
from IPython.display import display, Image, HTML

plots = sorted(PLOTS_DIR.glob('serie_*.png'))
print(f'{len(plots)} gráficos encontrados em {PLOTS_DIR}/')
```

```python
for plot in plots:
    acao_cod = plot.stem.replace('serie_', '')
    display(HTML(f'<h3 style="color:#1B3A5C; border-bottom: 2px solid #2E6DA4;">'
                 f'Ação {acao_cod}</h3>'))
    display(Image(str(plot), width=1000))
```

```python
# Gráficos de resumo global
resumo_plots = [
    ('resumo_modelos', 'Resumo — RMSE e MAE por Modelo'),
    ('vencedores', 'Modelos Vencedores'),
    ('distribuicao_rmse', 'Distribuição de RMSE por Modelo'),
]

for nome_plot, titulo in resumo_plots:
    p = PLOTS_DIR / f'{nome_plot}.png'
    if p.exists():
        display(HTML(f'<h3 style="color:#1B3A5C">{titulo}</h3>'))
        display(Image(str(p), width=900))
```
